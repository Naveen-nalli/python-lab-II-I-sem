#Counting no.of vowels and consonents
#Checking substring is present in a given string

```python
a=input()
b=a.lower()
v="aeiou"
count=0
for i in b:
  if i in v:
    count+=1
print(f"No.of vowels {count}")
print(f"No.of Consonents {len(b)-count}")
```
