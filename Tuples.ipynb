Creation of Tuple and Difference between Tuple and List



```python
import sys
L=[]
T=()
print(sys.getsizeof(L))
print(sys.getsizeof(T))   # List default Size is greater than Tuple
```

Opeartions On Tuple

```python
t1=(1,2,3)
t2=(4,5,6)
print(t1+t2) #Concatenation
print(t2*3)  #Repetition
print(3 in t1) #Membership
print(7 not in t2)
print(t1[2])  #Indexing
print(t1[0:3])
print(t2[::-1])  #Slicing
```

Functions:

```python
t3=(7,8,9)
t4=(10,11,12)
print(len(t3))  #Length
print(max(t3))  #Maximum element
print(min(t3))  #Minimum element
print(sorted(t3))  #Sorted Tuple
print(tuple(zip(t3,t4)))
```

Methods on Tuple:

```python
t1=(1,1,3,2,2,3)
print(t1.count(3))
print(t1.index(1))
```

# To find No.of duplicates in a list

```python
L=[int(i) for i in input().split()]
L1=[]
for i in L:
  if(L.count(i)>1):
    L1.append(i)
s=set(L1)
print(len(s))
```

# Print common elements between two Lists

```python
L1=[int(i) for i in input().split()]
L2=[int(i) for i in input().split()]
S1=set(L1)
S2=set(L2)
print(S1&S2)

```

# To find missing numbers in the range of 1 to n

```python
n=int(input())
L=[int(i) for i in input().split()]
num=set(range(1,n+1))
S=set(L)
print(num-S)
```

# To print unique elements in a list

```python
L=[int(i) for i in input().split()]
L1=[]
for i in L:
  if(L.count(i)==1):
    L1.append(i)
print(L1,len(L1))
```
